```python
import pandas as pd

df = pd.read_csv("../data/cleaned_data.csv")
df['Datetime'] = pd.to_datetime(df['Datetime'])
df.set_index('Datetime', inplace=True)

import matplotlib.pyplot as plt

daily = df['Global_active_power'].resample('D').mean()

daily.plot(figsize=(12,5))
plt.title("Daily Average Power Consumption")
plt.xlabel("Date")
plt.ylabel("Power")
plt.savefig("../images/daily_trend1.png")
plt.show()
```

```python
monthly = df['Global_active_power'].resample('ME').mean()

monthly.plot(figsize=(12,5))
plt.title("Monthly Average Power Consumption")
plt.savefig("../images/daily_trend2.png")
plt.show()
```

```python
df['hour'] = df.index.hour

hourly = df.groupby('hour')['Global_active_power'].mean()
hourly.plot(kind='bar', figsize=(10,5))
plt.title("Average Power Usage by Hour")
plt.savefig("../images/daily_trend3.png")
plt.show()
```

```python
df['day'] = df.index.dayofweek

weekday = df[df['day'] < 5]['Global_active_power']
weekend = df[df['day'] >= 5]['Global_active_power']

print("Weekday avg:", weekday.mean())
print("Weekend avg:", weekend.mean())
```

```python
import seaborn as sns

sns.histplot(df['Global_active_power'], bins=50, kde=True)
plt.title("Power Consumption Distribution")
plt.savefig("../images/daily_trend4.png")
plt.show()
```

```python
corr = df.corr()

sns.heatmap(corr, cmap='coolwarm')
plt.title("Feature Correlation")
plt.savefig("../images/daily_trend5.png")
plt.show()
```
